```python
######## ONTOLOGY ANNOTATOR...............................................

from rdflib import Graph, Namespace, URIRef, Literal
from rdflib.namespace import RDF, RDFS, OWL
import os
import time


# ============================================================
# CONFIGURATION
# ============================================================

ONTOLOGY_PATH = r"D:\BORUE\BiOnt.owl"

BASE_IRI = "http://localhost:3030/BiOnt.owl#"

BIONT = Namespace(BASE_IRI)

LANGUAGE_IN = BIONT.Language_In
STR_VAL = BIONT.strVal
URDUPHONE_CODE = BIONT.urduPhoneCode


# ============================================================
# GENERATE URDUPHONE CODE
# each word of a multiword strVal is encoded separately.
# ============================================================

def generate_urduphone_code(value):

    value = str(value).strip()

    if not value:
        return ""

    result = []

    # Same idea as Java:
    # value.split("\\s+")
    for word in value.split():

        word = word.strip()

        if not word:
            continue

        # ORIGINAL researchers' UrduPhone() function
        encoded = UrduPhone(word)

        if encoded is None:
            continue

        encoded = str(encoded).strip()

        if encoded:
            result.append(encoded)

    return " ".join(result)


# ============================================================
# COLLECT Language_In AND ALL ITS SUBCLASSES
#
# ============================================================

def collect_subclasses(graph, parent, result):

    if parent in result:
        return

    result.add(parent)

    for child in graph.subjects(
            RDFS.subClassOf,
            parent):

        if isinstance(child, URIRef):

            collect_subclasses(
                graph,
                child,
                result
            )


# ============================================================
# MAIN ANNOTATOR
# ============================================================

def annotate_urduphone():

    start_time = time.time()

    print()
    print("==========================================")
    print("URDUPHONE ONTOLOGY ANNOTATOR")
    print("==========================================")
    print()

    # --------------------------------------------------------
    # Safety checks
    # --------------------------------------------------------

    if not os.path.exists(ONTOLOGY_PATH):

        raise FileNotFoundError(
            "Ontology not found:\n" +
            ONTOLOGY_PATH
        )


    if "UrduPhone" not in globals():

        raise RuntimeError(
            "UrduPhone() is not defined.\n"
            "Run the researchers' UrduPhone code cell first."
        )


    # --------------------------------------------------------
    # Load ontology
    # --------------------------------------------------------

    graph = Graph()

    print("Loading ontology...")

    graph.parse(
        ONTOLOGY_PATH,
        format="xml"
    )

    print("Ontology loaded successfully.")
    print("Triples:", len(graph))
    print()


    # ========================================================
    # IMPORTANT: PRESERVE BiOnt PREFIX
    # ========================================================

    graph.bind(
        "BiOnt",
        BIONT,
        override=True,
        replace=True
    )


    # --------------------------------------------------------
    # Declare urduPhoneCode as OWL datatype property
    # --------------------------------------------------------

    graph.add(
        (
            URDUPHONE_CODE,
            RDF.type,
            OWL.DatatypeProperty
        )
    )


    print(
        "Data property:",
        URDUPHONE_CODE
    )


    # --------------------------------------------------------
    # Collect Language_In hierarchy
    # --------------------------------------------------------

    language_classes = set()

    collect_subclasses(
        graph,
        LANGUAGE_IN,
        language_classes
    )


    print()
    print(
        "Language classes found:",
        len(language_classes)
    )


    # --------------------------------------------------------
    # Find individuals whose asserted rdf:type is one of
    # Language_In or its subclasses.
    #
    # This mirrors your Java Soundex annotator.
    # --------------------------------------------------------

    language_individuals = set()

    for cls in language_classes:

        for individual in graph.subjects(
                RDF.type,
                cls):

            if isinstance(individual, URIRef):

                language_individuals.add(
                    individual
                )


    print(
        "Language individuals found:",
        len(language_individuals)
    )

    print()
    print("Processing...")
    print()


    processed = 0
    annotated = 0
    errors = 0
    no_strval = 0
    old_codes_removed = 0


    # --------------------------------------------------------
    # Process every Language_In individual
    # --------------------------------------------------------

    for individual in language_individuals:

        processed += 1


        # ----------------------------------------------------
        # Remove OLD urduPhoneCode assertion(s)
        # ----------------------------------------------------

        old_values = list(
            graph.objects(
                individual,
                URDUPHONE_CODE
            )
        )


        for old_value in old_values:

            graph.remove(
                (
                    individual,
                    URDUPHONE_CODE,
                    old_value
                )
            )

            old_codes_removed += 1


        # ----------------------------------------------------
        # Read strVal
        # ----------------------------------------------------

        str_values = list(
            graph.objects(
                individual,
                STR_VAL
            )
        )


        if not str_values:

            no_strval += 1
            continue


        # ----------------------------------------------------
        # Same behavior as Java SoundexAnnotator:
        # process each strVal literal.
        # ----------------------------------------------------

        for str_literal in str_values:

            value = str(
                str_literal
            ).strip()


            if not value:
                continue


            try:

                encoded_value = (
                    generate_urduphone_code(
                        value
                    )
                )


                if not encoded_value:
                    continue


                # ============================================
                # ADD:
                #
                # BiOnt:individual
                #     BiOnt:urduPhoneCode
                #     "CODE"
                # ============================================

                graph.add(
                    (
                        individual,
                        URDUPHONE_CODE,
                        Literal(encoded_value)
                    )
                )


                annotated += 1


                print(
                    value,
                    "-->",
                    encoded_value
                )


            except Exception as e:

                errors += 1

                print()
                print(
                    "ERROR:",
                    value
                )

                print(
                    "Individual:",
                    individual
                )

                print(
                    "Reason:",
                    e
                )

                print()


        if processed % 100 == 0:

            print()
            print(
                "Processed:",
                processed,
                "/",
                len(language_individuals)
            )

            print()


    # ========================================================
    # VERIFY BEFORE SAVING
    # ========================================================

    actual_assertions = sum(
        1
        for _ in graph.triples(
            (
                None,
                URDUPHONE_CODE,
                None
            )
        )
    )


    print()
    print("==========================================")
    print("PRE-SAVE VERIFICATION")
    print("==========================================")

    print(
        "Individuals processed:",
        processed
    )

    print(
        "Annotations generated:",
        annotated
    )

    print(
        "Actual urduPhoneCode assertions:",
        actual_assertions
    )

    print(
        "Old urduPhoneCode removed:",
        old_codes_removed
    )

    print(
        "Individuals without strVal:",
        no_strval
    )

    print(
        "Encoding errors:",
        errors
    )


    # --------------------------------------------------------
    # Critical safety check
    #
    # Do NOT overwrite ontology if nothing was annotated.
    # --------------------------------------------------------

    if actual_assertions == 0:

        raise RuntimeError(
            "\nZERO urduPhoneCode assertions were generated.\n"
            "ONTOLOGY HAS NOT BEEN SAVED."
        )


    # ========================================================
    # SAVE DIRECTLY BACK TO BiOnt.owl
    # ========================================================

    print()
    print("Saving ontology...")


    graph.serialize(
        destination=ONTOLOGY_PATH,
        format="pretty-xml"
    )


    # ========================================================
    # RELOAD AND VERIFY
    # ========================================================

    print(
        "Reloading saved ontology for verification..."
    )


    verification_graph = Graph()

    verification_graph.parse(
        ONTOLOGY_PATH,
        format="xml"
    )


    saved_assertions = sum(
        1
        for _ in verification_graph.triples(
            (
                None,
                URDUPHONE_CODE,
                None
            )
        )
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    print()
    print("==========================================")
    print("URDUPHONE ANNOTATION COMPLETE")
    print("==========================================")

    print(
        "Language individuals:",
        len(language_individuals)
    )

    print(
        "urduPhoneCode assertions saved:",
        saved_assertions
    )

    print(
        "Encoding errors:",
        errors
    )

    print(
        "Execution time:",
        round(elapsed, 2),
        "seconds"
    )

    print()

    print(
        "Ontology:",
        ONTOLOGY_PATH
    )

    print()

    print(
        "Property:",
        "BiOnt:urduPhoneCode"
    )


# ============================================================
# RUN
# ============================================================

annotate_urduphone()
```

```python
import os
import time
import openpyxl

from rdflib import Graph, Namespace, URIRef
from rdflib.namespace import RDF, RDFS


# ============================================================
# URDUPHONE EXCEL MATCHER
#
# Python counterpart of XXXExcelMatcher.java
# THIS WILL GENERATE URDUPHONE_RESULT.XLSX
# THIS WORK IS IN PYTHON SINCE THE ORIGINAL IMPLEMENTATION IS IN PYTHON
#
# IMPORTANT:
#   - BiOnt.owl is READ ONLY
#   - Nothing is written to the ontology
#   - Existing urduPhoneCode annotations are used
#   - Original researchers' UrduPhone() encodes queries
#   - ALL matching strVal values are returned
#   - NO Top-K
#   - NO ranking
#
# Excel:
#   Columns 1-6  = spelling variants
#   Column 7     = BORUE
#   Column 8     = Count
#   Columns 9-14 = UrduPhone results
# ============================================================


# ============================================================
# CONFIGURATION
# ============================================================

ONTOLOGY_PATH = r"F:\Eclipse\EclipseHome\BiOnt.owl"

EXCEL_PATH = (
    r"F:\Eclipse\EclipseHome\DS_TestBench"
    r"\TestBenchFinalCount.xlsx"
)

OUTPUT_PATH = (
    r"F:\Eclipse\EclipseHome\DS_TestBench"
    r"\UrduPhone_Result.xlsx"
)


BASE_IRI = "http://localhost:3030/BiOnt.owl#"

BIONT = Namespace(BASE_IRI)

STR_VAL = BIONT.strVal
URDUPHONE_CODE = BIONT.urduPhoneCode
LANGUAGE_IN = BIONT.Language_In


# ============================================================
# COLLECT Language_In + SUBCLASSES
#
# Python counterpart of:
#
# collectSubClasses(...)
#
# from SoundexExcelMatcher.java
# ============================================================

def collect_subclasses(graph, parent, result):

    if parent in result:
        return

    result.add(parent)

    for child in graph.subjects(
            RDFS.subClassOf,
            parent):

        if isinstance(child, URIRef):

            collect_subclasses(
                graph,
                child,
                result
            )


# ============================================================
# BUILD URDUPHONE INDEX
#
# Equivalent to:
#
# buildSoundexIndex(...)
#
# Java:
#
#     soundex code -> List<strVal>
#
# Python:
#
#     UrduPhone code -> list[strVal]
#
# IMPORTANT:
# Stored multiword codes are split exactly like the Java
# Soundex matcher splits soundexMeta on whitespace.
# ============================================================

def build_urduphone_index(
        graph,
        language_classes):

    index = {}

    count = 0


    # --------------------------------------------------------
    # Find all individuals belonging to Language_In hierarchy
    # --------------------------------------------------------

    language_individuals = set()

    for cls in language_classes:

        for individual in graph.subjects(
                RDF.type,
                cls):

            if isinstance(individual, URIRef):

                language_individuals.add(
                    individual
                )


    print(
        "Language individuals found:",
        len(language_individuals)
    )

    print()


    # --------------------------------------------------------
    # Build:
    #
    # urduPhoneCode -> strVal list
    # --------------------------------------------------------

    for individual in language_individuals:


        # ----------------------------------------------------
        # Read strVal
        # ----------------------------------------------------

        str_values = [
            str(value).strip()
            for value in graph.objects(
                individual,
                STR_VAL
            )
            if str(value).strip()
        ]


        # ----------------------------------------------------
        # Read urduPhoneCode
        # ----------------------------------------------------

        urduphone_values = [
            str(value).strip()
            for value in graph.objects(
                individual,
                URDUPHONE_CODE
            )
            if str(value).strip()
        ]


        # Same behaviour as Java matcher:
        #
        # if(strValues.isEmpty()
        #       ||
        #    soundexValues.isEmpty())
        #
        #     continue;

        if (
            not str_values
            or
            not urduphone_values
        ):
            continue


        # Java uses first strVal
        original_value = str_values[0]


        # Java uses first stored phonetic property
        stored_code_value = urduphone_values[0]


        # ----------------------------------------------------
        # EXACTLY mirrors:
        #
        # soundexValues.get(0).split("\\s+")
        #
        # This matters for multiword ontology entries.
        # ----------------------------------------------------

        codes = stored_code_value.split()


        for code in codes:

            code = code.strip()

            if not code:
                continue


            if code not in index:

                index[code] = []


            # Same duplicate protection as Java:
            #
            # if(!index.get(code)
            #          .contains(originalValue))
            #
            #     add(originalValue)

            if original_value not in index[code]:

                index[code].append(
                    original_value
                )


        count += 1


        if count % 500 == 0:

            print(
                "Individuals indexed:",
                count
            )


    return index


# ============================================================
# ENCODE QUERY
#
# IMPORTANT:
# We use the ORIGINAL researchers' UrduPhone() function.
#
# This corresponds to:
#
#     Soundex encoder = new Soundex();
#     String code = encoder.encode(value);
#
# in SoundexExcelMatcher.java.
# ============================================================

def encode_query(value):

    value = str(value).strip()

    if not value:
        return ""


    # --------------------------------------------------------
    # SoundexExcelMatcher encodes the complete Excel cell
    # with one call:
    #
    # encoder.encode(value)
    #
    # Therefore for UrduPhone we likewise make ONE call to
    # the original UrduPhone function for the query value.
    # --------------------------------------------------------

    return str(
        UrduPhone(value)
    ).strip()


# ============================================================
# PROCESS EXCEL
#
# Equivalent to:
#
# processExcel(...)
#
# from SoundexExcelMatcher.java
# ============================================================

def process_excel(
        excel_path,
        output_path,
        urduphone_index):


    print()
    print("Loading Excel:")
    print(excel_path)
    print()


    workbook = openpyxl.load_workbook(
        excel_path
    )


    sheet = workbook.worksheets[0]


    processed_rows = 0


    # --------------------------------------------------------
    # Java:
    #
    # for(Row row : sheet)
    #
    #     if(row.getRowNum() == 0)
    #         continue;
    #
    # Python therefore starts at row 2.
    # --------------------------------------------------------

    for row_number in range(
            2,
            sheet.max_row + 1):


        # ----------------------------------------------------
        # First SIX columns:
        #
        # A-F
        #
        # Python/openpyxl is 1-based.
        # ----------------------------------------------------

        for i in range(6):

            input_column = i + 1

            # Java:
            #
            # row.getCell(i + 8)
            #
            # Java POI is zero-based:
            #
            # 8 -> Excel column 9 = I
            #
            # Therefore:
            #
            # Variant 1 -> I
            # Variant 2 -> J
            # Variant 3 -> K
            # Variant 4 -> L
            # Variant 5 -> M
            # Variant 6 -> N

            output_column = i + 9


            input_cell = sheet.cell(
                row=row_number,
                column=input_column
            )


            output_cell = sheet.cell(
                row=row_number,
                column=output_column
            )


            # ------------------------------------------------
            # Clear any previous result in output cell
            # ------------------------------------------------

            output_cell.value = None


            if input_cell.value is None:
                continue


            value = str(
                input_cell.value
            ).strip()


            if not value:
                continue


            # ------------------------------------------------
            # Generate query UrduPhone code
            # ------------------------------------------------

            try:

                code = encode_query(
                    value
                )

            except Exception as e:

                print()
                print(
                    "ERROR encoding query:"
                )

                print(
                    "Excel row:",
                    row_number
                )

                print(
                    "Column:",
                    input_column
                )

                print(
                    "Value:",
                    value
                )

                print(
                    "Reason:",
                    e
                )

                print()

                continue


            # ------------------------------------------------
            # Exact code lookup
            # ------------------------------------------------

            matches = urduphone_index.get(
                code
            )


            # ------------------------------------------------
            # Write ALL matches.
            #
            # NO ranking.
            # NO Top-K.
            #
            # Same format as SoundexExcelMatcher:
            #
            # word1, word2, word3
            # ------------------------------------------------

            if matches:

                result_text = ", ".join(
                    matches
                )


                # --------------------------------------------
                # Excel maximum cell length protection.
                # --------------------------------------------

                if len(result_text) > 32767:

                    raise ValueError(
                        "\nExcel 32,767 character limit "
                        "exceeded.\n"
                        f"Row: {row_number}\n"
                        f"Query: {value}\n"
                        f"UrduPhone code: {code}\n"
                        f"Matches: {len(matches)}\n"
                        f"Characters: {len(result_text)}"
                    )


                output_cell.value = result_text


        processed_rows += 1


        if processed_rows % 500 == 0:

            print(
                "Excel rows processed:",
                processed_rows
            )


    # --------------------------------------------------------
    # Save NEW Excel file.
    #
    # Ontology is NEVER saved or modified.
    # --------------------------------------------------------

    print()
    print("Saving results...")


    workbook.save(
        output_path
    )


    workbook.close()


# ============================================================
# MAIN
# ============================================================

def main():

    start_time = time.time()


    print()
    print("==============================================")
    print("URDUPHONE EXCEL MATCHER")
    print("==============================================")
    print()


    # --------------------------------------------------------
    # Safety checks
    # --------------------------------------------------------

    if "UrduPhone" not in globals():

        raise RuntimeError(
            "UrduPhone() is not defined.\n"
            "Run the researchers' original UrduPhone "
            "code cell first."
        )


    if not os.path.exists(
            ONTOLOGY_PATH):

        raise FileNotFoundError(
            "Ontology not found:\n"
            + ONTOLOGY_PATH
        )


    if not os.path.exists(
            EXCEL_PATH):

        raise FileNotFoundError(
            "Excel file not found:\n"
            + EXCEL_PATH
        )


    # --------------------------------------------------------
    # Load ontology READ-ONLY
    # --------------------------------------------------------

    print("Loading ontology:")
    print(ONTOLOGY_PATH)
    print()


    graph = Graph()


    graph.parse(
        ONTOLOGY_PATH,
        format="xml"
    )


    print(
        "Ontology loaded successfully"
    )

    print(
        "Ontology triples:",
        len(graph)
    )


    # --------------------------------------------------------
    # Language_In hierarchy
    # --------------------------------------------------------

    language_classes = set()


    collect_subclasses(
        graph,
        LANGUAGE_IN,
        language_classes
    )


    print(
        "Language classes found:",
        len(language_classes)
    )


    # --------------------------------------------------------
    # Build UrduPhone index from ALREADY ANNOTATED ontology
    # --------------------------------------------------------

    urduphone_index = (
        build_urduphone_index(
            graph,
            language_classes
        )
    )


    print()
    print(
        "UrduPhone index size:",
        len(urduphone_index)
    )


    # --------------------------------------------------------
    # Critical verification
    # --------------------------------------------------------

    if len(urduphone_index) == 0:

        raise RuntimeError(
            "UrduPhone index is EMPTY.\n"
            "No usable urduPhoneCode annotations "
            "were found in the ontology."
        )


    # --------------------------------------------------------
    # Process TestBench
    # --------------------------------------------------------

    process_excel(
        EXCEL_PATH,
        OUTPUT_PATH,
        urduphone_index
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    print()
    print("==============================================")
    print("PROCESSING COMPLETED SUCCESSFULLY")
    print("==============================================")

    print(
        "Output file:"
    )

    print(
        OUTPUT_PATH
    )

    print()

    print(
        "Execution time:",
        round(elapsed, 2),
        "seconds"
    )


# ============================================================
# RUN
# ============================================================

main()
```
